```python
import numpy as np
```

```python
# Load projections from CSV
projections = np.loadtxt("processing_data/projections/projections_14x25.csv", delimiter=",")
```

```python
# use scipy to find peaks in the projections
from scipy.signal import find_peaks

for i, projection in enumerate(projections):

    peaks, _ = find_peaks(projection, height=0, threshold=0, distance=13)


    print("Peaks found at indices:", peaks)
    print("Number of peaks:", len(peaks))
    print("Peak shape", peaks.shape)

    # Export the peaks to a CSV file
    np.savetxt("processing_data/peaks/14x25/peaks_roi" + f"_{i+1}" + ".csv", peaks, delimiter=",")
```
